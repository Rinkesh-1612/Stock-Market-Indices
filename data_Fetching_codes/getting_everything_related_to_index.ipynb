```python
import yfinance as yf
import pandas as pd

# Set pandas to display more rows/columns if needed
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 120)

# --- Configuration ---
TICKER_SYMBOL = "^NSEI"

def explore_ticker_data(symbol):
    """
    Fetches and prints all available information for a given ticker symbol
    from the yfinance library.
    """
    print("=" * 70)
    print(f"🕵️  Exploring all available data for ticker: {symbol}")
    print("=" * 70)

    try:
        # 1. Create a Ticker object
        ticker = yf.Ticker(symbol)

        # 2. The '.info' Dictionary (Most comprehensive metadata)
        # ========================================================
        print("\n--- 1. Ticker.info (Metadata and Key Info) ---\n")
        info = ticker.info
        if info:
            # Pretty print the dictionary key-by-key
            for key, value in info.items():
                print(f"{key:<30}: {value}")
        else:
            print("No .info data found.")

        # 3. Historical Market Data
        # ========================================================
        print("\n\n--- 2. Ticker.history() (Price Data) ---\n")
        # Fetching last 1 year of data as a sample
        hist = ticker.history(period="1y")
        if not hist.empty:
            print("Sample of historical data (first 5 and last 5 rows):")
            print(hist.head())
            print("...")
            print(hist.tail())
        else:
            print("No historical data found for the specified period.")

        # 4. Corporate Actions (Dividends & Splits)
        # ========================================================
        print("\n\n--- 3. Ticker.actions (Dividends & Splits) ---\n")
        actions = ticker.actions
        if not actions.empty:
            print(actions)
        else:
            print("No dividends or splits found. (This is expected for an index).")

        # 5. Financial Statements (For Stocks, not Indices)
        # ========================================================
        print("\n\n--- 4. Financial Statements (Usually Empty for Indices) ---\n")
        print("Note: Indices are mathematical constructs, not companies, so they don't have financial statements.")
        
        print("\n--- 4a. ticker.financials ---")
        print(ticker.financials)
        
        print("\n--- 4b. ticker.balance_sheet ---")
        print(ticker.balance_sheet)
        
        print("\n--- 4c. ticker.cashflow ---")
        print(ticker.cashflow)
        
        # 6. Options Data
        # ========================================================
        print("\n\n--- 5. Ticker.options (Option Expiration Dates) ---\n")
        try:
            options = ticker.options
            if options:
                print("Available option expiration dates:")
                print(options)
            else:
                print("No options data available for this index.")
        except Exception as e:
            print(f"Could not fetch options data. Error: {e}")

        # 7. News Articles
        # ========================================================
        print("\n\n--- 6. Ticker.news (Recent News Articles) ---\n")
        news = ticker.news
        if news:
            print(f"Found {len(news)} news articles:")
            for i, article in enumerate(news):
                print(f"  {i+1}. {article['title']} ({article['publisher']})")
                print(f"     Link: {article['link']}")
        else:
            print("No recent news found.")

    except Exception as e:
        print(f"\nAN ERROR OCCURRED: Could not fetch data for {symbol}. It might be an invalid ticker.")
        print(f"Error details: {e}")

    print("\n" + "=" * 70)
    print("Exploration Complete.")
    print("=" * 70)


if __name__ == "__main__":
    explore_ticker_data(TICKER_SYMBOL)
```

```python
import yfinance as yf
import pandas as pd
import json

# Set pandas to display more rows/columns for better readability
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', 10)
pd.set_option('display.width', 120)

# --- Configuration ---
TICKER_SYMBOL = "MSFT" # Microsoft Corp.

def explore_company_data(symbol):
    """
    Fetches and prints all available information for a given company stock ticker
    from the yfinance library.
    """
    print("=" * 70)
    print(f"🕵️  Exploring all available data for company ticker: {symbol}")
    print("=" * 70)

    try:
        # 1. Create a Ticker object
        ticker = yf.Ticker(symbol)

        # 2. The '.info' Dictionary (Extremely rich for companies)
        # ========================================================
        print("\n--- 1. Ticker.info (Company Profile & Financial Ratios) ---\n")
        info = ticker.info
        if info and info.get('longBusinessSummary'):
            # Print the business summary first as it's often long
            print("--- Business Summary ---")
            print(info['longBusinessSummary'])
            print("\n--- Other Key Info ---")
            # Pretty print the rest of the dictionary
            for key, value in info.items():
                if key != 'longBusinessSummary':
                    print(f"{key:<30}: {value}")
        elif info:
             for key, value in info.items():
                print(f"{key:<30}: {value}")
        else:
            print("No .info data found.")

        # 3. Historical Market Data (Same as index, but more meaningful volume)
        # ========================================================
        print("\n\n--- 2. Ticker.history() (Price Data) ---\n")
        hist = ticker.history(period="1y")
        if not hist.empty:
            print("Sample of historical data (first 5 and last 5 rows):")
            print(hist.head())
            print("...")
            print(hist.tail())
        else:
            print("No historical data found.")

        # 4. Corporate Actions (CRITICAL for companies)
        # ========================================================
        print("\n\n--- 3. Ticker.actions (Dividends & Splits) ---\n")
        actions = ticker.actions
        if not actions.empty:
            print("Recent corporate actions (sample):")
            print(actions.tail(10)) # Show the 10 most recent actions
        else:
            print("No dividends or splits found.")

        # 5. Financial Statements (The core of company analysis)
        # ========================================================
        print("\n\n--- 4. Financial Statements (Annual) ---\n")
        print("--- 4a. Income Statement (Financials) ---")
        print(ticker.financials)
        print("\n--- 4b. Balance Sheet ---")
        print(ticker.balance_sheet)
        print("\n--- 4c. Cash Flow Statement ---")
        print(ticker.cashflow)
        
        # You can also get QUARTERLY data
        # print("\n--- Quarterly Income Statement ---")
        # print(ticker.quarterly_financials)
        
        # 6. Shareholder Information (Who owns the company?)
        # ========================================================
        print("\n\n--- 5. Shareholder Information ---\n")
        print("--- 5a. Major Holders ---")
        print(ticker.major_holders)
        print("\n--- 5b. Institutional Holders ---")
        print(ticker.institutional_holders.head(10)) # Show top 10

        # 7. Analyst Recommendations
        # ========================================================
        print("\n\n--- 6. Analyst Recommendations ---\n")
        recommendations = ticker.recommendations
        if recommendations is not None and not recommendations.empty:
            print("Recent analyst recommendations (sample):")
            print(recommendations.tail(10))
        else:
            print("No recommendation data found.")
            
        # 8. Company Calendar (Earnings Dates, etc.)
        # ========================================================
        print("\n\n--- 7. Company Calendar (Upcoming Events) ---\n")
        try:
            cal = ticker.calendar
            if cal is not None and not cal.empty:
                print(cal)
            else:
                 print("No calendar data found.")
        except Exception as e:
            print(f"Could not fetch calendar data. Error: {e}")
            
        # 9. News Articles
        # ========================================================
        print("\n\n--- 8. Ticker.news (Recent News Articles) ---\n")
        news = ticker.news
        if news:
            print(f"Found {len(news)} news articles (showing first 5):")
            for i, article in enumerate(news[:5]):
                print(f"  {i+1}. {article['title']} ({article['publisher']})")
        else:
            print("No recent news found.")

    except Exception as e:
        print(f"\nAN ERROR OCCURRED: Could not fetch data for {symbol}.")
        print(f"Error details: {e}")

    print("\n" + "=" * 70)
    print("Exploration Complete.")
    print("=" * 70)


if __name__ == "__main__":
    explore_company_data(TICKER_SYMBOL)
```
